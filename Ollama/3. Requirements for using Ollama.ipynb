# Requirements for Using Ollama
1. **Operating system**: Work on macOS, Windows, or Linux
2. **Hardware**: A laptop/PC with at least 8GB RAM (more = smoother)
3. **Storage Space**: Enough free disk space (models can be 3-15 GB each)
4. **Internet Connection (first time only)**: Needed to download Ollama and models.
5. **Basic Command Line Knowledge**: Knowing how to run simple terminal/command prompt commands.
6. **(Optional) GPU**- Not required, but speeds up performance if available
