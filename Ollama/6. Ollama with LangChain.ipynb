```python
# For Chat Purpose
from langchain_ollama import ChatOllama
```

```python
! ollama pull llama3.2:1b
```

```python
llm = ChatOllama(
    model='llama3.2:1b',
    temperature=0,
)
```

```python
response = llm.invoke("Is Your lie in april good movie?")
```
