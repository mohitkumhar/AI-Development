# What is Generative AI

**Generative AI** refers to a class of artificial intelligence models that can create new content - such as text, images, audio, code or video - that resembles human-created data.

> -  LLMs based apps like ChatGPT  
> - TTS models like ElevenLabs  
> - Diffusion models for images  
> - Code generating LLMs like CodeLLama  
> - Video gen models like Sora  

# Generative AI vs Traditional AI

- **Traditional AI** is about finding patterns in data and giving predictions.
---
- **Generative AI** is about learning the distribution (nature) of data so that it can generate a new sample from it.
    - The best part of Generative AI is that it can mimic humans.
