# Single Workflow Using LangGraph

```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
# define state
class BMIState(TypedDict):

    weight_kg: float
    height_m: float
    bmi: float
```

```python
def bmi_calculator(state: BMIState) -> BMIState:
    weight = state["weight_kg"]
    height = state["height_m"]

    bmi = weight / (height ** 2)

    state["bmi"] = round(bmi, 3)

    return state
```

```python
# define the grapg
graph = StateGraph(BMIState)

# add nodes to your graph
graph.add_node("calculate_bmi", bmi_calculator)

# add edges to your graph
graph.add_edge(START, "calculate_bmi")
graph.add_edge("calculate_bmi", END)

# compile the graph
workflow = graph.compile()
```

```python
workflow
```

```python
start_state = {"weight_kg": 12, "height_m": 1.9}
final_state = workflow.invoke(start_state)
```
